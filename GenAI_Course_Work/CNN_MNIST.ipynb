```python
#Load the data
from tensorflow.keras.datasets import fashion_mnist

((trainX, trainY), (testX, testY)) = fashion_mnist.load_data()
```

```python
print('Train: X=%s, y=%s' % (trainX.shape, trainY.shape))
print('Test: X=%s, y=%s' % (testX.shape, testY.shape))
```

```python

lables_train = np.zeros((60000,10 ))
for i in range(60000):
    lables_train[i][trainY[i]] = 1

data_train = trainX.reshape(60000,28,28,1)

lables_test = np.zeros((10000,10 ))
for i in range(10000):
    lables_test[i][testY[i]] = 1    
data_test = testX.reshape(10000,28,28,1)
```

```python
print(data_train.shape)
print(lables_train.shape)
print(data_test.shape)
print(lables_test.shape)    
```

```python
# Data normalization
data_train_norm = np.array(data_train / 255.0)
data_test_norm = np.array(data_test / 255.0)       

print(data_train_norm.shape)

print(data_test_norm.shape)
 
```

```python
print(data_train_norm.min(), data_train_norm.max())
print(data_test_norm.min(), data_test_norm.max())
```

```python
print(data_train_norm.dtype)
print(data_test_norm.dtype)
print(data_train_norm[0, :5, :5, 0])
```

```python
# build model
def build_model():
  model = Sequential()
  model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)))
  model.add(layers.MaxPooling2D((2, 2)))
  model.add(layers.Conv2D(64, (3, 3), activation='relu'))
  model.add(layers.MaxPooling2D((2, 2)))
  model.add(layers.Conv2D(64, (3, 3), activation='relu'))
  
  model.add(layers.Flatten())
  model.add(layers.Dense(64, activation='relu'))
  model.add(layers.Dense(10, activation='softmax'))

  compile = model.compile(optimizer='adam',
                loss='categorical_crossentropy',
                metrics=['accuracy'])
  
  return model
```

```python
model = build_model()
model.summary()
```

```python
# train the model
Epochs = 20
history = model.fit(
    data_train_norm, lables_train, epochs=Epochs, verbose=1, batch_size=100)
```

```python
test_loss, test_acc = model.evaluate(data_test_norm, lables_test, verbose=0)
print('\nTest accuracy:', int(test_acc * 100))
```
